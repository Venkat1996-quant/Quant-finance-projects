```python
import pandas as pd
from sklearn.linear_model import LinearRegression
#step(1) create data
data={'today':[100,102,101,103],'tomorrow':[102,101,103,104]}
df=pd.DataFrame(data)
#step(2) inputs
X=df[['today']]
#step(3)outputs
y=df['tomorrow']
#step(4) create model
model=LinearRegression()
#step(5) train model
model.fit(X, y)
#step(6) predict
predicted=model.predict(X)
print(predicted)
```
